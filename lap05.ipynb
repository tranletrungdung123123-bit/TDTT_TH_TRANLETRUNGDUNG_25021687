```python
from math import *
def prime(a):
  if a<2: return False
  for i in range(2,isqrt(a)+1):
      if a%i==0:
        return False
  return True
x=int(input())
print(prime(x))
```

```python
def pf(n):
  a=0
  for i in range(1,n):
    if n%i==0:
      a+=i
  if a==n: return True
  else: return False
x = int(input())
print(pf(x))
```

```python
s = input().strip()

if '/' in s:
    a, b = s.split('/')
    print("%.2f" % (float(a) / float(b)))
elif '*' in s:
    a, b = s.split('*')
    print("%.2f" % (float(a) * float(b)))
elif '+' in s:
    a, b = s.split('+')
    print("%.2f" % (float(a) + float(b)))
elif '-' in s:
    a, b = s.split('-')
    print("%.2f" % (float(a) - float(b)))
```

```python
def hamming(x,y):
  list1=[]
  list2=[]
  while x!=0:
    list1.append(x%2)
    x//=2
  while y!=0:
    list2.append(y%2)
    y//=2
  list1=list1[::-1]
  list2=list2[::-1]
  if len(list1)<len(list2):
    list1=[0]*(len(list2)-len(list1))+list1
  else:
    list2=[0]*(len(list1)-len(list2))+list2
  count=0
  for i in range(len(list1)):
    if list1[i]!=list2[i]:
      count+=1
  return count
a,b=map(int,input().split())
print(hamming(a,b))
```

```python
def anhxa(n):
    list1=[]
    list2=[]
    for i in n:
        if i not in list1:
            list1.append(i)
        list2.append(list1.index(i))
    return list2
def check(a,b):
    return anhxa(a)==anhxa(b)
a,b=map(str,input().split())
check(a,b)

```
