```python
#W7A1
def binary_search(a,l,r,x):
    res=-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            res=m
            return res
        elif a[m]>x:
            r=m-1
        else:l=m+1
    return res
arr = list(map(int, input().split()))
target = int(input())
left = 0
right = len(arr) - 1
print(binary_search(arr, left, right, target))

```

```python
#W7A3
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        swapped= False
        for j in range(n-1-i):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
                swapped=True
        if not swapped:
            break
    return arr
def selection_sort(a):
    n = len(a)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if a[j] < a[min_idx]:
                min_idx = j
        a[i], a[min_idx] = a[min_idx], a[i]
    return a
```

```python
#W7A6
def flatten(lst):
    result=[]
    for k in range(len(lst)):
        if k%2==0:
            result=result+lst[k]
        else:
            result=result+lst[k][::-1]
    return result
lst=eval(input())
print(flatten(lst))
```

```python
#W7A7
def length_of_lis(nums):
    if not nums:
        return 0
    
    dp = [1] * len(nums)
    
    for i in range(len(nums)):
        for j in range(i):
            if nums[i] > nums[j]:
                dp[i] = max(dp[i], dp[j] + 1)
                
    return max(dp)

nums=eval(input())
print(length_of_lis(nums))
```

```python
#W7A8
def len_x(x):
    return x[1] - x[0] + 1

def solve(arr, qs):
    arr.sort(key=len_x)
    ans = []
    
    for q in qs:
        val = -1
        for l, r in arr:
            if l <= q <= r:
                val = r - l + 1
                break
        ans.append(val)
        
    return ans

arr = eval(input())
qs = eval(input())

print(solve(arr, qs))
```

```python
#W7A9
def solve(arr):
    ans = []
    n = len(arr)
    
    for i in range(n):
        c = 0
        for j in range(i + 1, n):
            if arr[j] < arr[i]:
                c += 1
        ans.append(c)
        
    return ans

arr = eval(input())
print(solve(arr))
```

```python
#W7A10
import bisect

def max_subarray_mod(nums, m):
    seen_prefix = []
    max_mod = 0
    curr_sum = 0
    
    for x in nums:
        curr_sum = (curr_sum + x) % m
        max_mod = max(max_mod, curr_sum)
        
        idx = bisect.bisect_right(seen_prefix, curr_sum)
        
        if idx < len(seen_prefix):
            prev_sum = seen_prefix[idx]
            max_mod = max(max_mod, (curr_sum - prev_sum + m) % m)
            
        bisect.insort(seen_prefix, curr_sum)
        
    return max_mod

nums = eval(input())
m = int(input())
print(max_subarray_mod(nums, m))
```

```python
#W7A13
def findLHS(nums):
    counts = {}
    for x in nums:
        if x in counts:
            counts[x] += 1
        else:
            counts[x] = 1
            
    max_len = 0
    for x in counts:
        if x + 1 in counts:
            current_len = counts[x] + counts[x+1]
            if current_len > max_len:
                max_len = current_len
                
    return max_len

nums = eval(input())
print(findLHS(nums))
```
