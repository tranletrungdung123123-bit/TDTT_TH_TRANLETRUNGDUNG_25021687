```python
class nextday:
    def __init__(self, d, m, y):
        self.d = d
        self.m = m
        self.y = y

    def namnhuan(self):
        return (self.y % 400 == 0) or (self.y % 4 == 0 and self.y % 100 != 0)

    def ngaytoida(self):
        if self.m in [1, 3, 5, 7, 8, 10, 12]:
            return 31
        elif self.m in [4, 6, 9, 11]:
            return 30
        elif self.m == 2:
            return 29 if self.namnhuan() else 28
        return 0

    def check(self):
        if self.y < 1 or self.m < 1 or self.m > 12:
            return False
        if 1 <= self.d <= self.ngaytoida():
            return True
        return False

    def ngaytieptheo(self):
        d_next = self.d + 1
        m_next = self.m
        y_next = self.y

        if d_next > self.ngaytoida():
            d_next = 1
            m_next += 1
            if m_next > 12:
                m_next = 1
                y_next += 1
        return nextday(d_next, m_next, y_next)

    def __str__(self):
        return f"{self.d:02d}/{self.m:02d}/{self.y:04d}"

try:
    s = input().strip()
    parts = s.split('/')
    if len(parts) == 3:
        d = int(parts[0])
        m = int(parts[1])
        y = int(parts[2])

        current_date = nextday(d, m, y)

        if current_date.check():
            ket_qua = current_date.ngaytieptheo()
            print(ket_qua)
        else:
            print("INVALID")
    else:
        print("INVALID")
except:
    print("INVALID")
```

```python
#W8A3
import math

class khonggian:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def la_goc_toa_do(self):
        return self.x == 0 and self.y == 0

    def nam_tren_truc_tung(self):
        return self.x == 0

    def nam_tren_truc_hoanh(self):
        return self.y == 0

    def tinh_khoang_cach_goc(self):
        return math.sqrt(self.x**2 + self.y**2)

try:
    dong_nhap = input().split()
    if len(dong_nhap) == 2:
        x_m = float(dong_nhap[0])
        y_m = float(dong_nhap[1])

        m = khonggian(x_m, y_m)

        if m.la_goc_toa_do():
            print("M la goc toa do")
        
        if m.nam_tren_truc_tung():
            print("M nam tren truc tung")
        
        if m.nam_tren_truc_hoanh():
            print("M nam tren truc hoanh")
        
        if not m.nam_tren_truc_tung() and not m.nam_tren_truc_hoanh():
            print("M khong nam tren cac truc toa do")

        d = m.tinh_khoang_cach_goc()
        print(f"{d:.2f}")
except:
    print("INVALID")
```

```python
class Calculator():
    def __init__(self, n1, n2):
        self.n1 = n1
        self.n2 = n2

    def add(self):
        return self.n1 + self.n2

    def subtract(self):
        return self.n1 - self.n2

    def product(self):
        return self.n1 * self.n2

    def divide(self):
        if self.n2 == 0:
            return "INVALID (Loi chia cho 0)"
        return self.n1 / self.n2

    def power(self):
        return self.n1 ** self.n2

    def mod(self):
        if self.n2 == 0:
            return "INVALID (Loi chia cho 0)"
        return self.n1 % self.n2

    def set_numbers(self, n1, n2):
        self.n1 = n1
        self.n2 = n2

print("Nhap hai so ban dau:")
a = float(input("So thu nhat: "))
b = float(input("So thu hai: "))
my_calc = Calculator(a, b)

while True:
    print("\n--- MENU MAY TINH ---")
    print("1. Cong (add)")
    print("2. Tru (subtract)")
    print("3. Nhan (product)")
    print("4. Chia (divide)")
    print("5. Luy thua (power)")
    print("6. Chia du (mod)")
    print("7. Cap nhat 2 so moi (set_numbers)")
    
    choice = input("Moi ban chon (1-7): ")

    if choice == '1':
        print("Ket qua:", my_calc.add())
    elif choice == '2':
        print("Ket qua:", my_calc.subtract())
    elif choice == '3':
        print("Ket qua:", my_calc.product())
    elif choice == '4':
        print("Ket qua:", my_calc.divide())
    elif choice == '5':
        print("Ket qua:", my_calc.power())
    elif choice == '6':
        print("Ket qua:", my_calc.mod())
    elif choice == '7':
        new_a = float(input("Nhap so thu nhat moi: "))
        new_b = float(input("Nhap so thu hai moi: "))
        my_calc.set_numbers(new_a, new_b)
        print("Da cap nhat so moi!")
    else:
        print("Lua chon khong hop le.")

    exit_ask = input("Ban co muon thoat? (Go 'yes' de thoat): ")
    if exit_ask.lower() == "yes":
        print("Cam on ban da su dung!")
        break
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def them_san_pham(self, name, price):
        product = {"name": name, "price": price}
        self.items.append(product)
        print(f"Da them '{name}' vao gio hang.")

    def xoa_san_pham(self, name):
        tim_thay = False
        for item in self.items:
            if item["name"].lower() == name.lower():
                self.items.remove(item)
                tim_thay = True
                print(f"Da xoa '{name}' khoi gio hang.")
                break
        if not tim_thay:
            print(f"Khong tim thay san pham ten '{name}'.")

    def kiem_tra_rong(self):
        return len(self.items) == 0

    def tinh_tong_tien(self):
        tong = sum(item["price"] for item in self.items)
        return tong

    def hien_thi_gio_hang(self):
        if self.kiem_tra_rong():
            print("Gio hang dang trong.")
        else:
            print("--- Cac san pham trong gio ---")
            for i, item in enumerate(self.items, 1):
                print(f"{i}. {item['name']} - Gia: {item['price']:.2f}")

    def xoa_toan_bo(self):
        self.items = []
        print("Da xoa sach gio hang.")

gio_hang = ShoppingCart()

while True:
    print("\n--- MENU QUAN LY GIO HANG ---")
    print("2. Them san pham moi")
    print("3. Xoa san pham theo ten")
    print("4. Kiem tra gio hang rong")
    print("5. Tinh tong tien")
    print("6. Hien thi toan bo gio hang")
    print("7. Xoa sach gio hang")
    
    lua_chon = input("Nhap lua chon cua ban (2-7): ")

    if lua_chon == "2":
        ten = input("Nhap ten san pham: ")
        gia = float(input("Nhap gia san pham: "))
        gio_hang.them_san_pham(ten, gia)
    elif lua_chon == "3":
        ten = input("Nhap ten san pham can xoa: ")
        gio_hang.xoa_san_pham(ten)
    elif lua_chon == "4":
        if gio_hang.kiem_tra_rong():
            print("Gio hang dang trong.")
        else:
            print("Gio hang hien co san pham.")
    elif lua_chon == "5":
        print(f"Tong tien trong gio hang la: {gio_hang.tinh_tong_tien():.2f}")
    elif lua_chon == "6":
        gio_hang.hien_thi_gio_hang()
    elif lua_chon == "7":
        gio_hang.xoa_toan_bo()
    else:
        print("Lua chon khong hop le, vui long chon tu 2 toi 7.")

    thoat = input("Co muon thoat khoi chuong trinh khong? (Go 'yes' de thoat): ")
    if thoat.lower() == "yes":
        print("Tam biet!")
        break
```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h

    def area(self):
        return self.w * self.h

    def perimeter(self):
        return 2 * (self.w + self.h)

    def scale(self, k):
        self.w = self.w * k
        self.h = self.h * k

try:
    dong_nhap = input().split()
    if len(dong_nhap) == 3:
        w = float(dong_nhap[0])
        h = float(dong_nhap[1])
        k = float(dong_nhap[2])

        hinh_chu_nhat = Rectangle(w, h)
        
        hinh_chu_nhat.scale(k)
        
        s = hinh_chu_nhat.area()
        p = hinh_chu_nhat.perimeter()
        
        print(f"{s:.2f} {p:.2f}")
except:
    print("Du lieu nhap vao khong hop le")
```

```python
import math

class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("INVALID")
        
        ucln = math.gcd(num, den)
        self.num = num // ucln
        self.den = den // ucln
        
        if self.den < 0:
            self.num = -self.num
            self.den = -self.den

    def add(self, other):
        tu_moi = self.num * other.den + other.num * self.den
        mau_moi = self.den * other.den
        return Fraction(tu_moi, mau_moi)

    def sub(self, other):
        tu_moi = self.num * other.den - other.num * self.den
        mau_moi = self.den * other.den
        return Fraction(tu_moi, mau_moi)

    def mul(self, other):
        tu_moi = self.num * other.num
        mau_moi = self.den * other.den
        return Fraction(tu_moi, mau_moi)

    def div(self, other):
        tu_moi = self.num * other.den
        mau_moi = self.den * other.num
        return Fraction(tu_moi, mau_moi)

    def __str__(self):
        return f"{self.num}/{self.den}"

try:
    dong_nhap = input().split()
    if len(dong_nhap) == 5:
        a = int(dong_nhap[0])
        b = int(dong_nhap[1])
        phep_toan = dong_nhap[2]
        c = int(dong_nhap[3])
        d = int(dong_nhap[4])

        ps1 = Fraction(a, b)
        ps2 = Fraction(c, d)

        if phep_toan == '+':
            ket_qua = ps1.add(ps2)
        elif phep_toan == '-':
            ket_qua = ps1.sub(ps2)
        elif phep_toan == '*':
            ket_qua = ps1.mul(ps2)
        elif phep_toan == '/':
            ket_qua = ps1.div(ps2)
        
        print(ket_qua)
except:
    print("INVALID")
```

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = float(balance)

    def deposit(self, x):
        self.balance += x

    def withdraw(self, x):
        if x <= self.balance:
            self.balance -= x

    def __str__(self):
        return f"{self.owner}: {self.balance:.2f}"

try:
    input_line1 = input().split()
    if len(input_line1) >= 2:
        owner_name = input_line1[0]
        initial_balance = float(input_line1[1])
        
        my_account = BankAccount(owner_name, initial_balance)
        
        n = int(input())
        
        for i in range(n):
            command_line = input().split()
            action = command_line[0].upper()
            amount = float(command_line[1])
            
            if action == "DEPOSIT":
                my_account.deposit(amount)
            elif action == "WITHDRAW":
                my_account.withdraw(amount)
        
        print(f"{my_account.balance:.2f}")

except:
    print("INVALID")
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(float(score))

    def avg(self):
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        diem_tb = self.avg()
        if diem_tb >= 8:
            return "Excellent"
        elif diem_tb >= 6.5:
            return "Good"
        elif diem_tb >= 5:
            return "Average"
        else:
            return "Poor"

try:
    name_input = input().strip()
    n = int(input().strip())
    
    hoc_sinh = Student(name_input)
    
    for i in range(n):
        line = input().split()
        if len(line) >= 2:
            mon_hoc = line[0]
            diem = float(line[1])
            hoc_sinh.add_score(mon_hoc, diem)
            
    diem_trung_binh = hoc_sinh.avg()
    xep_loai = hoc_sinh.rank()
    
    print(f"{hoc_sinh.name} {diem_trung_binh:.2f} {xep_loai}")

except:
    pass
```

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = int(year)

class Library:
    def __init__(self):
        self.books = []

    def add(self, book):
        self.books.append(book)

    def count_by_author(self, name):
        count = 0
        for b in self.books:
            if b.author == name:
                count += 1
        return count

    def find_by_year(self, y):
        count = 0
        for b in self.books:
            if b.year == int(y):
                count += 1
        return count

try:
    n = int(input().strip())
    lib = Library()
    
    for _ in range(n):
        line = input().split(maxsplit=1)
        cmd = line[0]
        args = line[1] if len(line) > 1 else ""
        
        if cmd == "ADD":
            parts = args.split(";")
            if len(parts) == 3:
                lib.add(Book(parts[0], parts[1], parts[2]))
        elif cmd == "COUNT":
            print(lib.count_by_author(args))
        elif cmd == "COUNTYEAR":
            print(lib.find_by_year(args))
except:
    pass
```

```python
import math

class Point:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    
    def __sub__(self, other):
        return Vector(
            self.x - other.x,
            self.y - other.y,
            self.z - other.z
        )

class Vector:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    
    def dot(self, other):
        return (self.x * other.x +
                self.y * other.y +
                self.z * other.z)
    
    def cross(self, other):
        return Vector(
            self.y * other.z - self.z * other.y,
            self.z * other.x - self.x * other.z,
            self.x * other.y - self.y * other.x
        )
    
    def magnitude(self):
        return math.sqrt(self.x**2 + self.y**2 + self.z**2)


# Đọc 4 điểm (mỗi điểm 1 dòng)
points = []
for _ in range(4):
    line = input().strip()
    if not line:
        continue
    x, y, z = map(float, line.split())
    points.append(Point(x, y, z))

A, B, C, D = points

# Vector AB, AC cho mặt phẳng ABC
AB = B - A
AC = C - A

# Vector BC, BD cho mặt phẳng BCD
BC = C - B
BD = D - B

# Vector pháp tuyến n1 của mặt phẳng ABC
n1 = AB.cross(AC)

# Vector pháp tuyến n2 của mặt phẳng BCD
n2 = BC.cross(BD)

# Tính cos phi = |n1 · n2| / (|n1| * |n2|)
dot_prod = abs(n1.dot(n2))
mag_n1 = n1.magnitude()
mag_n2 = n2.magnitude()

if mag_n1 == 0 or mag_n2 == 0:
    # Một trong hai mặt phẳng suy biến (3 điểm thẳng hàng)
    phi_deg = 0.00
else:
    cos_phi = dot_prod / (mag_n1 * mag_n2)
    # Giới hạn cos_phi trong [-1, 1] để tránh lỗi số học
    cos_phi = max(min(cos_phi, 1.0), -1.0)
    phi_rad = math.acos(cos_phi)
    phi_deg = math.degrees(phi_rad)

# In kết quả với 2 chữ số thập phân
print(f"{phi_deg:.2f}")
```

```python
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.realPart = initRealPart
        self.imaginaryPart = initImaginaryPart
    
    def multiply(self, other):
        real = (self.realPart * other.realPart -
                self.imaginaryPart * other.imaginaryPart)
        imag = (self.realPart * other.imaginaryPart +
                self.imaginaryPart * other.realPart)
        return ComplexNumber(real, imag)


try:
    a, b = map(float, input().split())
    c, d = map(float, input().split())
    
    z1 = ComplexNumber(a, b)
    z2 = ComplexNumber(c, d)
    
    result = z1.multiply(z2)
    
    # In với 2 chữ số thập phân
    print(f"{result.realPart:.2f} {result.imaginaryPart:.2f}")

except:
    pass
```

```python
import math

class Triangle:
    def __init__(self, a, b, c):
        self.a, self.b, self.c = a, b, c
    
    def area(self):
        if min(self.a, self.b, self.c) <= 0:
            return -1
        if self.a + self.b <= self.c or self.a + self.c <= self.b or self.b + self.c <= self.a:
            return -1
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))

t = Triangle(*map(int, input().split()))
res = t.area()
print("invalid" if res < 0 else f"{res:.2f}")
```
