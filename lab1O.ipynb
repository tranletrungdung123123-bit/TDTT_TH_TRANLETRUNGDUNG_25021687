```python

n = input()
s = 0
for i in n:
     s += (int(i))**2
try:
    seen = set()

    if s == 1:
        print(s)
    else:
        while s != 1:
            if s in seen:
                raise ValueError("Vòng lặp vô hạn")
            else:
                seen.add(s)
            snew = 0
            scopy = str(s)            #for i in str(s):
            for i in scopy:
                snew += (int(i))**2
            s = snew
    print("YES")
except ValueError:
    print("NO")
```
