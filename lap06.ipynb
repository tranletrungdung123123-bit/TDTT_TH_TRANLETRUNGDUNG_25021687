```python
#W6A5
nums = list(map(int, input().split()))
result = {
    'positives': 0,
    'negatives': 0,
    'zeros': 0
}
for x in nums:
    if x > 0:
        result['positives'] += 1
    elif x < 0:
        result['negatives'] += 1
    else:
        result['zeros'] += 1
print(result)
```

```python
#W6A9
def nhapdict():
    n=int(input())
    d={}
    for _ in range(n):
        k,v=input().split()
        d[k]=int(v)
    return d
d1=nhapdict()
d2=nhapdict()
for x in d2:
    if x not in d1:
        d1[x]=d2[x]
    else:
        d1[x]+=d2[x]
a=dict(sorted(d1.items()))
print(a)
```
